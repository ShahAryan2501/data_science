# QUESTION - 3.

```python
def linear_regression_predict(x, m, c):
    y = m * x + c
    return y

distances = [2, 5, 8]

for distance in distances:
    delivery_time = linear_regression_predict(distance, 4, 10)
    print("Distance:", distance, "km")
    print("Predicted delivery time:", delivery_time, "minutes")
```

# QUESTION - 4.

The slope

The slope tells us how much the predicted product price changes when the product rating 
increases by 1 point. 
For example, if the slope is ₹500, then a 1-point increase in rating means 
the predicted price increases by ₹500.

Intercept

The intercept is the predicted product price when the product rating is 0. 
It represents the starting value of the prediction when the input value is zero.
