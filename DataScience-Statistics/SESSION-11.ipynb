# QUESTION - 3.

```python
orders = [
    [10, 20, 15, 12],
    [15, 25, 10, 20],
    [20, 15, 25, 18]
]

total_orders = [0, 0, 0, 0]

for row in orders:
    for i in range(4):
        total_orders[i] += row[i]

print("Total orders for each dish:")

for i in range(4):
    print("Dish", i + 1, ":", total_orders[i])
```

# QUESION - 4.

```python
products = [
    [500, 10],
    [800, 15],
    [1200, 8],
    [300, 20],
    [1000, 12]
]

def calculate_revenue(price, quantity):
    return price * quantity

for i in range(len(products)):
    price = products[i][0]
    quantity = products[i][1]
    revenue = calculate_revenue(price, quantity)
    print("Product", i + 1, "Revenue:", revenue)
```
