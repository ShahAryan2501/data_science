# QUESTION - 2.

```python
from scipy.stats import binom

def bayes_posterior(prior, likelihood, evidence):
    return (prior * likelihood) / evidence

prior_premium = 0.10
likelihood_discount_premium = 0.80

prior_regular = 0.90
likelihood_discount_regular = 0.20

evidence = (prior_premium * likelihood_discount_premium) + (prior_regular * likelihood_discount_regular)

posterior = bayes_posterior(
    prior_premium,
    likelihood_discount_premium,
    evidence
)

print("Probability that restaurant is premium:", posterior)
print("Probability percentage:", posterior * 100)
```

```python
from scipy.stats import binom

def bayes_posterior(prior, likelihood, evidence):
    return (prior * likelihood) / evidence

prior = 0.20
likelihood = 0.70

prior_other = 0.80
likelihood_other = 0.30

evidence = (prior * likelihood) + (prior_other * likelihood_other)

posterior = bayes_posterior(
    prior,
    likelihood,
    evidence
)

print("Probability that restaurant is highly rated:", posterior)
print("Probability percentage:", posterior * 100)
```
