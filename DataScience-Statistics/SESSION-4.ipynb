# QUESTION - 1.

```python
import statistics
```

```python
orders = [12, 15, 17, 14, 13, 16, 200, 18, 14, 15]
mean = statistics.mean(orders)
print("mean :",mean)

median = statistics.median(orders)
print("median :",median)
print("here : mean > median.")
print("right_skewed distribution.")
```

# QUESTION - 2.

```python
followers = [5, 7, 8, 8, 9, 10, 12, 15, 95]
mean = statistics.mean(followers)
print("mean :",mean)

median = statistics.median(followers)
print("median :",round(median,2))
print("here : mean > median.")
print("right_skewed distribution.")
```

# QUESTION - 4.

```python
ratings = [3, 3, 4, 4, 4, 5, 5, 5, 5, 5]
mean = statistics.mean(ratings)
print("mean :",round(mean,2))

median = statistics.median(ratings)
print("median :",median)

if mean > median:
    print("right_skewed distribution.")

else:
    print("final answer :- left_skewed distribution.")
```
