# QUESTION - 1.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

data = np.random.normal(50, 10, 100)

mean = np.mean(data)
sd = np.std(data)

x = np.linspace(min(data), max(data), 100)
y = norm.pdf(x, mean, sd)

plt.plot(x, y)
plt.xlabel("Values")
plt.ylabel("Probability Density")
plt.title("Normal Distribution")
plt.show()
```

# QUESTION - 2.

```python
import numpy as np
from scipy import stats

steps = np.array([
    5000, 5200, 4800, 5100, 5300,
    4950, 5050, 5150, 4900, 5250,
    5000, 5100, 4950, 5200, 5050,
    5150, 5000, 5100, 4950, 5200
])

mean = np.mean(steps)
median = np.median(steps)
mode = stats.mode(steps, keepdims=False)

print("Mean:", mean)
print("Median:", median)
print("Mode:", mode.mode)

if abs(mean - median) < 200 and abs(median - mode.mode) < 200:
    print("Mean, median and mode are approximately equal.")
else:
    print("Mean, median and mode are not approximately equal.")
```

# QUESTION - 3.

```python
import numpy as np
from scipy import stats

scores = np.random.normal(70, 15, 200)

lower = 70 - 15
upper = 70 + 15

count = np.sum((scores >= lower) & (scores <= upper))

percentage = (count / 200) * 100

print("Number of players:", 200)
print("Players scoring between 55 and 85:", count)
print("Percentage:", percentage)

expected = 0.68 * 200

print("Expected number according to 68-95-99 rule:", expected)
```
