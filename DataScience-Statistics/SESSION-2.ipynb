# QUESTION - 4.

```python
import statistics
```

```python
print("the statistics of channel_A is :-")
channel_A = [100, 120, 110, 105, 5000]
mean = statistics.mean(channel_A)
print("the mean of this list is :",mean)
median = statistics.median(channel_A)
print("the median of this list is :",median)
mode = statistics.mode(channel_A)
print("the mode of this list is :",mode)

print("\nthe statistics of channel_B is :-")
channel_B = [200, 220, 210, 205, 215]
mean = statistics.mean(channel_B)
print("the mean of this list is :",mean)
median = statistics.median(channel_B)
print("the median of this list is :",median)
mode = statistics.mode(channel_B)
print("the mode of this list is :",mode)


print("\nthe statistics of channel_C is :-")
channel_C = [100, 150, 100, 200, 100]
mean = statistics.mean(channel_C)
print("the mean of this list is :",mean)
median = statistics.median(channel_C)
print("the median of this list is :",median)
mode = statistics.mode(channel_C)
print("the mode of this list is :",mode)

print("\nChannel A is most affected by outliers because 5000 is an extreme value.")
```
