# QUESTION - 3.

We use **sampling** because Swiggy has a very large number of orders, 
so collecting and analyzing data from every single order would take a 
lot of **time, money, and resources**. By selecting a representative sample, 
we can analyze delivery times more quickly and efficiently while still getting 
useful information about the overall population.

```python
import statistics

delivery_times = [25, 30, 28, 35, 27, 32, 29, 31]

average_time = statistics.mean(delivery_times)

print("Sample Average Delivery Time:", average_time, "minutes")
```

# RANDOM SAMPLING.

```python
import random
import statistics

users = [f"user{i}" for i in range(1, 1001)]

sample = random.sample(users, 50)

print("Selected Users:")
print(sample)

bookings = [random.randint(1, 10) for i in range(50)]

print("Average Movie Bookings:", statistics.mean(bookings))
```

# STRATIFIEND SAMPLING.

```python
import random
import statistics

age_18_25 = [f"user{i}" for i in range(1, 401)]
age_26_40 = [f"user{i}" for i in range(401, 801)]
age_41_60 = [f"user{i}" for i in range(801, 1001)]

sample1 = random.sample(age_18_25, 10)
sample2 = random.sample(age_26_40, 10)
sample3 = random.sample(age_41_60, 10)

sample = sample1 + sample2 + sample3

print("Selected Users:")
print(sample)

bookings = [random.randint(1, 10) for i in range(30)]

print("Average Movie Bookings:", statistics.mean(bookings))
```
