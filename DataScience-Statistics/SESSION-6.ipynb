# QUESTION - 1.

```python
from scipy import stats

instagram = [60, 70, 55, 80, 65, 75, 90, 50, 85, 70]
youtube = [50, 60, 55, 65, 50, 60, 70, 45, 65, 55]

t_stat, p_value = stats.ttest_ind(instagram, youtube)

print("t-statistic:", t_stat)
print("p-value:", p_value)

if p_value < 0.05:
    print("Reject H0")
else:
    print("Fail to reject H0")
```

# QUESTION - 2.

```python
from scipy import stats

mean_A = 350
mean_B = 320

sd_A = 50
sd_B = 50

n_A = 30
n_B = 30

standard_error = ((sd_A ** 2 / n_A) + (sd_B ** 2 / n_B)) ** 0.5

t_stat = (mean_A - mean_B) / standard_error

df = n_A + n_B - 2

p_value = 2 * stats.t.sf(abs(t_stat), df)

print("t-statistic:", t_stat)
print("p-value:", p_value)

if p_value < 0.05:
    print("Reject H0")
else:
    print("Fail to reject H0")
```

# QUESTION - 3.

```python
from scipy import stats

control = [40, 45, 50, 42, 48, 44, 46, 41, 49, 45]
variant = [50, 55, 52, 60, 58, 54, 56, 51, 59, 57]

t_stat, p_value = stats.ttest_ind(control, variant)

print("t-statistic:", t_stat)
print("p-value:", p_value)

if p_value < 0.05:
    print("Reject H0")
    print("New dark mode has a significant effect.")
else:
    print("Fail to reject H0")
    print("New dark mode does not have a significant effect.")
```

```python
from scipy import stats

data = [
    [520, 480],
    [480, 520]
]

chi2, p_value, dof, expected = stats.chi2_contingency(data)

print("Chi-square:", chi2)
print("p-value:", p_value)

if p_value < 0.05:
    print("Reject H0")
    print("The difference is statistically significant.")
else:
    print("Fail to reject H0")
    print("The difference may be due to chance.")
```
