# QUESTION - 1.

```python
import statistics

def calculate_probability(event_count, total_outcomes):
    return event_count / total_outcomes

heads = 1
total_outcomes = 2

probability = calculate_probability(heads, total_outcomes)

print("Probability of getting Heads:", probability)
print("Probability percentage:", probability * 100)
```

# QUESTION - 2.

```python
import random
import statistics

results = []

for i in range(100):
    dice = random.randint(1, 6)
    results.append(dice)

six_count = results.count(6)

probability = six_count / len(results)

print("Number of times 6 appeared:", six_count)
print("Experimental probability of getting 6:", probability)
print("Probability percentage:", probability * 100)
```

# QUESTION - 3.

```python
import statistics

genres = [
    "pop", "rock", "pop", "hiphop", "jazz",
    "pop", "rock", "pop", "hiphop", "pop",
    "rock", "jazz", "pop", "pop", "rock",
    "hiphop", "pop", "jazz", "rock", "pop"
]

pop_count = genres.count("pop")
total_songs = len(genres)

probability = pop_count / total_songs

print("Number of Pop songs:", pop_count)
print("Total songs:", total_songs)
print("Probability of picking Pop:", probability)
print("Probability percentage:", probability * 100)
```

```python
import statistics

def probability_both(food_users, dessert_users):
    food_set = set(food_users)
    dessert_set = set(dessert_users)

    both_users = food_set.intersection(dessert_set)
    all_users = food_set.union(dessert_set)

    probability = len(both_users) / len(all_users)

    return probability

food_users = [1, 2, 3, 4, 5, 6, 7]
dessert_users = [3, 4, 5, 8, 9]

result = probability_both(food_users, dessert_users)

print("Probability of ordering both food and dessert:", result)
print("Probability percentage:", result * 100)
```
