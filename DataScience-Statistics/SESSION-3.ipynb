# QUESTION - 2.

```python
def calculate_standred_deviation(scores):
    a = statistics.mean(scores)
    print("the mean of this list is :",a)

    b = statistics.pstdev(scores)
    print("the population standred deviation is :",b)
    print("the population standred deviation is :",int(b))


calculate_standred_deviation([60, 70, 80, 90, 100])
```
