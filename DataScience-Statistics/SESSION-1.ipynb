# QUESTION - 4.

For a Spotify-like playlist recommendation system, the following statistical terms 
would be useful:

1. Mean

Mean represents the average value of the data.

Example: If a user listens to music for 20, 30, 40, 50, and 60 minutes, 
the mean listening time is 40 minutes.

2. Median

Median is the middle value when the data is arranged in order.

Example: For listening times 20, 30, 40, 50, and 60 minutes, the median is 40 minutes.

3. Mode

Mode is the value or category that occurs most frequently.

Example: If a user listens to Pop music more frequently than other genres, 

Pop is the mode and can be used for playlist recommendations.
