# QUESTION - 2.

```python
user_preferences = np.array([
    [5, 3, 2],
    [4, 5, 3],
    [2, 4, 5]
])


song_popularity = np.array([
    [4, 2, 3],
    [3, 5, 2],
    [2, 3, 5]
])


result_dot = np.dot(user_preferences, song_popularity)

print("Result using dot():")
print(result_dot)

result_matmul = np.matmul(user_preferences, song_popularity)

print("Result using matmul():")
print(result_matmul)

# Difference:
# For 2D matrices, np.dot() and np.matmul() give the same result.
# np.dot() is a general dot product function,
# while np.matmul() is specifically used for matrix multiplication.

# Main point:** 3 * 3 matrices ke case mein `dot()` aur `matmul()` ka result same hai. 
```

# QUESTION - 3.

```python
pixels = np.array([
    [120, 150, 200, 180],
    [90,  110, 160, 140],
    [200, 220, 180, 170],
    [130, 100, 190, 210]
])

print("Original pixel matrix:")
print(pixels)

transposed = pixels.T

print("\nTransposed matrix:")
print(transposed)

print("\nMean:", np.mean(pixels))
print("Median:", np.median(pixels))
print("Standard Deviation:", np.std(pixels))
print("Variance:", np.var(pixels))

```

# QUESTION - 4.

```python
matrix = np.random.randint(1,6,(3,3))
print("the 3 cross 3 matrix is :")
print(matrix)

determinant = np.linalg.det(matrix)
print("\n the determinant of this matrix is ;",determinant)

inverse = np.linalg.inv(matrix)
print("\n the inverse of this matrix is ;")
print(inverse)

eigvalues = np.linalg.eig(matrix)
print("\n the eigen vectors and eigen values is :")
print(eigvalues)
```

# QUESTION - 5.

```python
array = np.array([[11,12,33,44,77,88],[1,2,3,4,5,6]])
print("the array is :")
print(array)

array2 = np.reshape(array,(3,4))
print("\nthe reshape array is :")
print(array2)

array3 = array2.flatten()
print("\nthe flatten array is :")
print(array3)

part1, part2 = np.split(array3, 2)

print("\nPart 1:")
print(part1)

print("\nPart 2:")
print(part2)

stacked = np.vstack((part1, part2))

print("\nVertically stacked array:")
print(stacked)
```
