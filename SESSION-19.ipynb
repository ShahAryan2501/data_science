# QUESTION - 4.

```python
class FoodOrder:
    def __init__(self,restaurant_name,items,total_price):
        self.restaurant_name = restaurant_name
        self.items = items
        self.total_price = total_price

    def add_item(self,item_name,item_price):
        self.items.append(item_name)
        self.total_price = self.total_price + item_price

    def display(self):
        print("the restaurant name is :",self.restaurant_name)
        print("the item name is :",self.items)
        print("the total price is :",self.total_price)


x = FoodOrder("mayur restaurants.",["pizza","burger"],550)

x.add_item("garlic bread.",220)
x.add_item("cake",530)

x.display()
```
