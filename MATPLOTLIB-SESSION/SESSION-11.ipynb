```python
import matplotlib.pyplot as plt

# Data for Line Chart
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
temperature = [28, 30, 29, 32, 31]

# Data for Bar Chart
subjects = ["Python", "Java", "C++", "SQL"]
marks = [85, 72, 78, 90]

# Data for Scatter Chart
hours = [1, 2, 3, 4, 5, 6]
scores = [40, 45, 55, 60, 72, 85]

# Data for Pie Chart
categories = ["Food", "Travel", "Shopping", "Entertainment"]
expenses = [40, 25, 20, 15]

# Create 2x2 subplots
fig, ax = plt.subplots(2, 2, figsize=(10, 8))

# 1. Line Chart
ax[0, 0].plot(days, temperature, marker="o")
ax[0, 0].set_title("Temperature")
ax[0, 0].set_xlabel("Days")
ax[0, 0].set_ylabel("Temperature")

# 2. Bar Chart
ax[0, 1].bar(subjects, marks)
ax[0, 1].set_title("Student Marks")
ax[0, 1].set_xlabel("Subjects")
ax[0, 1].set_ylabel("Marks")

# 3. Scatter Chart
ax[1, 0].scatter(hours, scores)
ax[1, 0].set_title("Study Hours vs Scores")
ax[1, 0].set_xlabel("Study Hours")
ax[1, 0].set_ylabel("Scores")

# 4. Pie Chart
ax[1, 1].pie(expenses, labels=categories, autopct="%1.1f%%")
ax[1, 1].set_title("Monthly Expenses")

# Adjust spacing
plt.tight_layout()

# Display
plt.show()
```

```python
import matplotlib.pyplot as plt

services = ["Zomato", "Swiggy", "Dominos"]
delivery_time = [32, 38, 28]

plt.bar(services[0], delivery_time[0], color="red", edgecolor="black", linewidth=2)
plt.bar(services[1], delivery_time[1], color="blue", edgecolor="green", linewidth=3)
plt.bar(services[2], delivery_time[2], color="orange", edgecolor="purple", linewidth=1)

plt.title("Average Delivery Time")
plt.xlabel("Food Delivery Service")
plt.ylabel("Delivery Time (minutes)")

plt.show()
```

```python
import matplotlib.pyplot as plt

influencers = ["Virat", "Ronaldo", "MrBeast", "Selena", "CarryMinati"]
followers = [270, 650, 60, 430, 18]
daily_posts = [1.2, 1.8, 2.5, 1.5, 3.0]

fig, ax1 = plt.subplots()

ax1.plot(
    influencers,
    followers,
    color="blue",
    marker="o"
)

ax1.set_xlabel("Influencers")
ax1.set_ylabel("Followers (Millions)")

ax2 = ax1.twinx()

ax2.plot(
    influencers,
    daily_posts,
    color="red",
    marker="s"
)

ax2.set_ylabel("Average Daily Posts")

plt.title("Instagram Followers vs Daily Posts")

plt.show()
```

```python
import matplotlib.pyplot as plt

movies = ["Jawan", "Pathaan", "Animal", "Dangal", "KGF 2"]
tickets_sold = [950, 880, 820, 1100, 760]
imdb_rating = [7.0, 5.9, 6.1, 8.3, 8.3]

plt.scatter(imdb_rating, tickets_sold)

for i in range(len(movies)):
    plt.annotate(
        movies[i],
        (imdb_rating[i], tickets_sold[i]),
        xytext=(5, 5),
        textcoords="offset points"
    )

plt.xlabel("IMDb Rating")
plt.ylabel("Tickets Sold (Lakhs)")
plt.title("IMDb Rating vs Tickets Sold")

plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

sales = [120, 135, 150, 145, 160, 175,
         180, 190, 210, 350, 230, 200]

fig, ax = plt.subplots()

ax.plot(months, sales, marker="o")

ax.annotate(
    "Big Billion Days",
    xy=("Oct", 350),
    xytext=("Sep", 390),
    arrowprops=dict(arrowstyle="->")
)

ax.set_xlabel("Month")
ax.set_ylabel("Monthly Sales")
ax.set_title("Flipkart Monthly Sales")

plt.show()
```
