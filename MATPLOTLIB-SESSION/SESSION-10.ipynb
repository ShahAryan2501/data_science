# TASK - 1.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
steps = [4500, 6200, 5800, 7500, 8200, 10000, 6800]

plt.plot(days,steps,color="yellow",linewidth=2,alpha=0.5)
plt.xlabel("days")
plt.ylabel("steps")
plt.title("7 days steps data")
plt.savefig("steps_lineplot.png")
plt.show()
```

# TASK - 2.

```python
restaurants = [
    "Food Hub", "Spice Villa", "Pizza Point", "Urban Cafe",
    "Tasty Treat", "Royal Dine", "Green Leaf", "Food Corner",
    "Desi Rasoi", "The Cafe"
]

ratings = [4.2, 3.8, 4.5, 4.0, 3.6, 4.7, 4.1, 3.9, 4.3, 4.6]

prices = [350, 250, 600, 400, 200, 800, 300, 450, 280, 700]

plt.scatter(ratings,prices,color="blue",linewidth=2,alpha=0.5)

plt.title("restaurants data.")
plt.xlabel("restaurants ratings.")
plt.ylabel("food prices.")

plt.show()
```

# TASK - 3.

```python
platforms = ["Swiggy", "Zomato", "Dominos"]
orders = [120, 150, 90]

plt.bar(platforms,orders,color=["red","blue","yellow"])

plt.xlabel("platform name.")
plt.ylabel("number of orders.")
plt.title("platforms data.")

plt.legend(["orders"])

plt.show()
```

# TASK - 4.

```python
durations = [12, 25, 18, 35, 42, 15, 28, 50, 32, 20,
             45, 38, 22, 55, 30, 17, 40, 48, 26, 34]

plt.hist(durations,color="pink",bins=5,linewidth=2)

plt.title("Spotify Listening Session Duration.")
plt.xlabel("Listening Duration.")
plt.ylabel("Frequemcy.")
plt.legend("Listening Sessions.")
plt.grid()
plt.savefig("data.png")
plt.show()
```

# TASK - 5.

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

screen_time = [2.5, 3, 2, 4, 3.5, 5, 4.5]
posts_liked = [20, 35, 15, 40, 30, 55, 45]

fig, ax = plt.subplots(1, 2, figsize=(10, 4))

ax[0].plot(days, screen_time, marker="o",color="yellow")
ax[0].set_title("Instagram Screen Time")
ax[0].set_xlabel("Days")
ax[0].set_ylabel("Screen Time (Hours)")

ax[1].bar(days, posts_liked, color="skyblue")
ax[1].set_title("Posts Liked")
ax[1].set_xlabel("Days")
ax[1].set_ylabel("Number of Posts Liked")

plt.tight_layout()

plt.savefig("social_media_usage.png")
plt.show()
```
