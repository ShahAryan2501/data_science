```python
import pandas as pd

df = pd.read_csv("paytm_payments.csv")

df["payment_status"] = df["payment_status"].str.strip().str.lower()

df["payment_status"] = df["payment_status"].replace({
    "yes": 1,
    "y": 1,
    "no": 0,
    "n": 0
})

print(df)
```
