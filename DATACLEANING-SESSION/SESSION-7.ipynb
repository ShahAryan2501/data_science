# TASK - 1.

```python
import pandas as pd

df = pd.read_csv("zomato_ratings.csv")

Q1 = df["user_rating"].quantile(0.25)
Q3 = df["user_rating"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df["user_rating"] < lower) | (df["user_rating"] > upper)]

print("Outlier indices:")
print(outliers.index)
```

# TASK - 3.

```python
import pandas as pd

df = pd.read_csv("paytm_transactions.2csv")

lower = df["transaction_amount"].quantile(0.05)
upper = df["transaction_amount"].quantile(0.95)

df["transaction_amount"] = df["transaction_amount"].clip(lower, upper)

print(df["transaction_amount"].describe())
```

```python
import pandas as pd

df = pd.read_csv("spotify_users.csv")

df["is_premium"] = df["is_premium"].apply(
    lambda x: True if str(x).lower() in ["true", "1", "yes"] else False
)

print(df)
print(df.dtypes)
```
