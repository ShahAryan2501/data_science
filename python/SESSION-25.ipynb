# QUESTION - 1.

```python
import requests
import json

symbols = [
    "BTCUSDT",
    "ETHUSDT",
    "BNBUSDT",
    "XRPUSDT",
    "SOLUSDT",
    "DOGEUSDT",
    "ADAUSDT",
    "TRXUSDT",
    "AVAXUSDT",
    "LINKUSDT"
]

url = "https://api.binance.com/api/v3/ticker/24hr"

all_data = []

for symbol in symbols:

    params = {
        "symbol": symbol
    }

    response = requests.get(url, params=params)

    if response.status_code == 200:
        all_data.append(response.json())

    else:
        print("Error:", symbol, response.status_code)

with open("crypto_data.json", "w") as file:
    json.dump(all_data, file, indent=4)

print("Data saved successfully in crypto_data.json")
```

# QUESTION - 2.

```python
import json

def find_most_volatile_coin(data):

    coin = max(
        data,
        key=lambda x: abs(float(x["priceChangePercent"]))
    )

    return coin["symbol"]


with open("crypto_data.json", "r") as file:
    data = json.load(file)

result = find_most_volatile_coin(data)

print("Most Volatile Coin:", result)
```

# QUESTION - 3.

```python
import json

with open("crypto_data.json", "r") as file:
    data = json.load(file)

prices = []

for coin in data:
    prices.append(float(coin["lastPrice"]))

average_price = sum(prices) / len(prices)

print("Average Price:", average_price)

print("\nCoins below average price:")

for coin in data:

    price = float(coin["lastPrice"])

    if price < average_price:
        print(coin["symbol"], ":", price)
```

# QUESTION - 4.

```python
import json

def rank_coins_by_volume(data):

    sorted_data = sorted(
        data,
        key=lambda x: float(x["volume"]),
        reverse=True
    )

    print("Top 5 Coins by Volume:")
    print("-------------------------")

    for rank, coin in enumerate(sorted_data[:5], start=1):

        print(
            rank,
            coin["symbol"],
            "Volume:",
            coin["volume"]
        )


with open("crypto_data.json", "r") as file:
    data = json.load(file)

rank_coins_by_volume(data)
```

```python
import requests
import json
import time

symbols = [
    "BTCUSDT",
    "ETHUSDT",
    "BNBUSDT",
    "XRPUSDT",
    "SOLUSDT",
    "DOGEUSDT",
    "ADAUSDT",
    "TRXUSDT",
    "AVAXUSDT",
    "LINKUSDT"
]

def fetch_data():

    url = "https://api.binance.com/api/v3/ticker/24hr"

    all_data = []

    for symbol in symbols:

        try:
            response = requests.get(
                url,
                params={"symbol": symbol},
                timeout=10
            )

            if response.status_code == 200:
                all_data.append(response.json())

            elif response.status_code == 429:
                print("Rate limit reached. Please wait.")

            else:
                print("API Error:", response.status_code)

        except requests.exceptions.RequestException:
            print("Network Error")

    if len(all_data) > 0:

        with open("crypto_data.json", "w") as file:
            json.dump(all_data, file, indent=4)

        print("Data updated successfully!")

fetch_data()

print("Program finished.")
```
