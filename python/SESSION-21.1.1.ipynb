# QUESTION - 1.

```python
import requests

url = "https://jsonplaceholder.typicode.com/posts"

data = {
    "title": "My First Post",
    "body": "This is my post body",
    "userId": 1
}

response = requests.post(url, json=data)

print("Status Code:", response.status_code)
print("JSON Data:", response.json())
```

```python
import requests

url = "https://jsonplaceholder.typicode.com/posts"

name = input("Enter playlist name: ")
print("the playlist name is :",name)
description = input("Enter playlist description: ")
print("the playlist descripion is :",description)

data = {
    "title": name,
    "body": description
}

response = requests.post(url, json=data)

result = response.json()

print("Playlist ID:", result["id"])
```

# QUESTION - 4.

```python
import requests
import csv

url = "https://jsonplaceholder.typicode.com/posts"

response = requests.get(url)

posts = response.json()

latest_5 = posts[-5:]

with open("posts.csv", "w", newline="") as file:
    writer = csv.writer(file)

    writer.writerow(["Title", "User ID"])

    for post in latest_5:
        writer.writerow([post["title"], post["userId"]])

print("Data saved successfully in posts.csv")
```

# QUESTION - 5.

```python
import requests
import json

url = "https://jsonplaceholder.typicode.com/posts"

response = requests.get(url)

posts = response.json()

latest_5 = posts[-5:]

with open("posts.json", "w") as file:
    json.dump(latest_5, file, indent=4)

print("Data saved successfully in posts.json")
```
