# QUESTION - 1.

```python
import requests
```

```python
url = "https://jsonplaceholder.typicode.com/posts"

a = requests.get(url)
print(a.status_code)
b = a.json()

for i in range(1):
    print("the first title is :",b[i]["title"])
```

# QUESTION - 2.

```python
url = "https://jsonplaceholder.typicode.com/posts"

data = {
    "title" : "My First Post",
    "body" : "Hello from Python!",
    "userId" : 101
}

x = requests.post(url,json=data)
print("the status code is :",x.status_code)
print("the new data is :",x.json())
```

# QUESTION - 3.

```python
url = "https://jsonplaceholder.typicode.com/users"

s = requests.get(url)
y = s.json()

for i in y:
    if i["email"].endswith(".org"):
        print(i["username"])
```
