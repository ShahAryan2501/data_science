# QUESTION - 1.

```python
song_titles = ['Shape Of You', 'Blinding Lights', 'Levitating', 'Senorita']
new_song_titles = list(map(lambda x : x.lower(),song_titles))
print("the all songs in lower letter is :",new_song_titles)
```

# QUESTION - 2.

```python
Zomato_restaurant_ratings = [4.2, 3.8, 4.5, 2.9, 3.5]
new_ratings = list(filter(lambda x : x > 4.0,Zomato_restaurant_ratings))
print("the new zomato ratings is above 4.0 :",new_ratings)
```

# QUESTION - 3.

```python
from functools import reduce
Flipkart_shopping_cart =  [499, 1299, 299, 799]
Total_sum_this_cart = reduce(lambda x,y : x+y,Flipkart_shopping_cart)
print("the total sum of this cart is :",Total_sum_this_cart)
```

# QUESTION - 4.

```python
numbers = [950,1500,25000,1200000]
new_numbers = list(map(lambda x : f"{x/1000000}M" if x >= 1000000 else f"{x/1000}K" if x >= 1000 else f"{x}",numbers))
print(new_numbers)
```
