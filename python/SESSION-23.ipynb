# QUESTION - 1.

```python
import requests

api_key = "YOUR_ACTUAL_API_KEY"

url = "https://api.openweathermap.org/data/2.5/weather"

params = {
    "q": "Ahmedabad",
    "appid": api_key,
    "units": "metric"
}

response = requests.get(url, params=params)

print("Status Code:", response.status_code)

if response.status_code == 200:
    data = response.json()

    print("City:", data["name"])
    print("Temperature:", data["main"]["temp"], "°C")
    print("Weather:", data["weather"][0]["description"])

else:
    print("Error:", response.json())
```

# QUESTION - 2.

```python
import requests
from datetime import datetime

url = "https://api.coingecko.com/api/v3/simple/price"

params = {
    "ids": "bitcoin,ethereum",
    "vs_currencies": "usd"
}

response = requests.get(url, params=params)

if response.status_code == 200:
    data = response.json()

    print("Date and Time:", datetime.now())
    print("Bitcoin Price: $", data["bitcoin"]["usd"])
    print("Ethereum Price: $", data["ethereum"]["usd"])

else:
    print("Error:", response.status_code)
```

# QUESTION - 3.

```python
import requests

api_key = "DEMO_KEY"

url = "https://api.nasa.gov/planetary/apod"

params = {
    "api_key": api_key
}

response = requests.get(url, params=params)

if response.status_code == 200:

    data = response.json()

    print("Title:", data["title"])
    print()
    print("Explanation:")
    print(data["explanation"])

    image_url = data["url"]

    image = requests.get(image_url)

    with open("nasa_apod.jpg", "wb") as file:
        file.write(image.content)

    print()
    print("Image saved as nasa_apod.jpg")

else:
    print("Error:", response.status_code)
```

# QUESTION - 4.

```python
import requests

url = "https://disease.sh/v3/covid-19/countries/India"

response = requests.get(url)

if response.status_code == 200:

    data = response.json()

    print("--------------------------------")
    print("       COVID-19 - INDIA")
    print("--------------------------------")
    print("Total Cases  :", data["cases"])
    print("Recovered    :", data["recovered"])
    print("Deaths       :", data["deaths"])
    print("--------------------------------")

else:
    print("Error:", response.status_code)
```

# QUESTION - 5.

```python
import requests

# OpenWeatherMap
weather_key = "YOUR_API_KEY"

weather_url = "https://api.openweathermap.org/data/2.5/weather"

weather_params = {
    "q": "Mumbai",
    "appid": weather_key,
    "units": "metric"
}

# CoinGecko
crypto_url = "https://api.coingecko.com/api/v3/simple/price"

crypto_params = {
    "ids": "bitcoin",
    "vs_currencies": "usd"
}

try:

    weather_response = requests.get(
        weather_url,
        params=weather_params
    )

    crypto_response = requests.get(
        crypto_url,
        params=crypto_params
    )

    if weather_response.status_code == 200:
        weather_data = weather_response.json()
        temperature = weather_data["main"]["temp"]

        print("Mumbai Temperature:", temperature, "°C")
    else:
        print("Weather API failed")

    if crypto_response.status_code == 200:
        crypto_data = crypto_response.json()
        bitcoin_price = crypto_data["bitcoin"]["usd"]

        print("Bitcoin Price: $", bitcoin_price)
    else:
        print("Bitcoin API failed")

except Exception as e:
    print("Something went wrong:", e)
```
