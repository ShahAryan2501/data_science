# QUESTION - 3.

```python
url = "https://jsonplaceholder.typicode.com/posts"

data = {
    "title" : "My Python Playlist",
    "userId" : 1,
    "body" : "My favorite songs"
}

x = requests.post(url,json=data)
print(x.json())
```

# QUESTION - 4.

```python
url = "https://jsonplaceholder.typicode.com/posts"

params = {
    "userId": 2
}

y = requests.get(url,params=params)
z = y.json()

for i in z:
    print(f"the only idsin this url is :",i["id"])
```
