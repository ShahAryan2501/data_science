# QUESTION - 3.

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def get_discounted_price(self):
        return self.price - (self.price * 10 / 100)


class Electronics(Product):
    def get_discounted_price(self):
        return self.price - (self.price * 20 / 100)


def show_final_price(item):
    print("Product Name :", item.name)
    print("Discounted Price :", item.get_discounted_price())


p = Product("Shoes", 2000)
e = Electronics("Laptop", 50000)

show_final_price(p)
show_final_price(e)
```

# QUESTION - 4.

```python
class Ticket:
    def __init__(self, price):
        self.price = price

    def get_final_price(self):
        return self.price


class PremiumTicket(Ticket):
    def get_final_price(self):
        return super().get_final_price() + 50


t1 = Ticket(300)
t2 = PremiumTicket(300)

print("Normal Ticket Price :", t1.get_final_price())
print("Premium Ticket Price :", t2.get_final_price())
```
