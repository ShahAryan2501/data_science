```python
import pandas as pd

mumbai = pd.DataFrame({
    "restaurant": ["Spice Hub", "Mumbai Masala", "Urban Cafe", "Tasty Bites", "Royal Kitchen"],
    "category": ["North Indian", "Indian", "Cafe", "Fast Food", "Chinese"],
    "average_cost_for_two": [700, 600, 500, 400, 800],
    "user_rating": [4.2, 4.0, 4.3, 3.8, 4.1]
})

mumbai.to_csv("zomato_mumbai.csv", index=False)

print("Mumbai CSV created")
```

```python
delhi = pd.DataFrame({
    "restaurant": ["Delhi Darbar", "Royal Dine", "Capital Cafe", "Food Junction", "Punjabi Tadka"],
    "category": ["North Indian", "Indian", "Cafe", "Fast Food", "North Indian"],
    "average_cost_for_two": [500, 550, 450, 350, 600],
    "user_rating": [4.0, 3.8, 4.1, 3.7, 4.2]
})

delhi.to_csv("zomato_delhi.csv", index=False)

print("Delhi CSV created")
```

```python
!pip install sweetviz
```

```python
import sweetviz as sv

report = sv.compare(
    [mumbai, "Mumbai"],
    [delhi, "Delhi"]
)

report.show_html("zomato_mumbai_vs_delhi.html")
```
