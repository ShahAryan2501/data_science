```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("flipcart_reviews.csv")

print(df.head())

print(df["rating"].value_counts().sort_index())

sns.countplot(data=df, x="rating")

plt.title("Number of Reviews by Rating")
plt.xlabel("Rating")
plt.ylabel("Number of Reviews")

plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("zomato_task3.csv")

sns.scatterplot(
    data=df,
    x="average_cost_for_two",
    y="user_rating"
)

plt.title("Average Cost vs User Rating")
plt.xlabel("Average Cost for Two")
plt.ylabel("User Rating")

plt.show()
```
