# QUESTION - 1.

```python
class Playlist:
    def __init__(self):
        self.__songs = []

    def add_song(self,song):
        self.__songs.append(song)

    def display_song(self):
        print("the playlist is :",self.__songs)

a = Playlist()

a.add_song("saiyaara")
a.add_song("saibo")
a.add_song("dhun")

a.display_song()
```

# QUESTION - 2.

```python
class Product:
    def __init__(self):
        self.__price = 0

    def set_price(self,price):
        self.__price = price

    def get_price(self):
        return self.__price

b = Product()

b.set_price(5000)

print("the product price is :",b.get_price())
```

```python
class Movie_Ratings:
    def __init__(self):
        self.__ratings = 0

    def set_ratings(self,rating):
        if 0 <= rating <= 10:
             self.__ratings = rating
             print("valid ratings.")
        else:
            print("invalidratings.")
            print("ratings must be between 0 to 10.")


    def get_ratings(self):
        return self.__ratings

c = Movie_Ratings()

c.set_ratings(8) 

print("the ratings is :",c.get_ratings())
```
