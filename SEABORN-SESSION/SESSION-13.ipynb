# TASK - 2.

```python
flights = sns.load_dataset("flights")
flights
```

```python
data = flights.pivot_table(
    index="month",
    columns="year",
    values="passengers"
)

sns.heatmap(data)

plt.title("Passengers by Month and Year")
plt.xlabel("Year")
plt.ylabel("Month")

plt.show()
```
