# TASK - 1.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
delivery_time = [32, 41, 28, 35, 45, 30, 52, 38, 29, 47,
                 33, 36, 42, 25, 39, 44, 31, 50, 27, 34,
                 40, 37, 48, 29, 43, 35, 26, 51, 32, 46,
                 38, 41, 30, 36, 49, 28, 33, 45, 39, 31,
                 42, 35, 47, 29, 40, 52, 34, 37, 44, 30]

data = pd.DataFrame(delivery_time)
data
```

```python
data.head()
```

```python
sns.histplot(data = data)
sns.despine()
plt.xlabel("Delivery time in minutes.")
plt.ylabel("number of orders.")
plt.title("zomato delivery time distribution.")

plt.show()
```

# TASK - 3.

```python
teams = ["CSK", "MI", "RCB", "KKR", "GT", "RR", "DC", "PBKS"]

scores = [175, 162, 188, 145, 170, 155, 192, 148,
          160, 180, 172, 135, 185, 150, 165, 178,
          155, 190, 145, 168, 175, 182, 158, 140,
          195, 160, 170, 150, 180, 165, 155, 185,
          172, 148, 190, 162, 175, 155, 180, 145]


df = pd.DataFrame({
    "Team": teams * 5,
    "Runs": scores
})
```

```python
df
```

```python
sns.violinplot(data=df,x="Team",y="Runs")
sns.set_style("darkgrid")
plt.xlabel("team name")
plt.ylabel("team scores")
plt.title("ipl data")

plt.show()
```
