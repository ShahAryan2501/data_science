```python
import pandas as pd
from sqlalchemy import create_engine

engine = create_engine("sqlite:///zomato.db")

data = {
    "id": [1, 2, 3, 4, 5],
    "name": ["Spice Hub", "Pizza Point", "Royal Kitchen", "Burger House", "South Spice"],
    "cuisine": ["Indian", "Italian", "Punjabi", "Fast Food", "South Indian"],
    "rating": [4.5, 4.2, 4.7, 4.0, 4.4],
    "city": ["Ahmedabad", "Ahmedabad", "Gandhinagar", "Ahmedabad", "Surat"]
}

restaurants = pd.DataFrame(data)

restaurants.to_sql("restaurants", engine, index=False, if_exists="replace")

df = pd.read_sql("restaurants", engine)

print(df.head())
```

# TASK - 2.

```python
movies_data = {
    "name": ["3 Idiots", "Dangal", "Jawan", "12th Fail", "Taare Zameen Par"],
    "rating": [8.4, 8.3, 7.0, 9.0, 8.3]
}

movies = pd.DataFrame(movies_data)

movies.to_sql("movies", engine, index=False, if_exists="replace")
```

```python
query = "SELECT name, rating FROM movies WHERE rating > 8"

df = pd.read_sql_query(query, engine)

print(df)
```

```python
orders = pd.DataFrame({
    "order_id": [101, 102, 103, 104, 105],
    "user_id": [1, 2, 3, 1, 5],
    "amount": [450, 700, 320, 550, 900]
})

orders.to_csv("orders.csv", index=False)
```

```python
import pandas as pd
from pathlib import Path

orders_path = Path("orders.csv")
users_path = Path("users.csv")

orders_df = pd.read_csv(orders_path)
users_df = pd.read_csv(users_path)

df = pd.merge(orders_df, users_df, on="user_id")

print(df)
```

```python
today_orders = pd.DataFrame({
    "order_id": [101, 102, 103],
    "item": ["Pizza", "Burger", "Biryani"],
    "price": [300, 200, 350]
})

yesterday_orders = pd.DataFrame({
    "order_id": [98, 99, 100],
    "item": ["Sandwich", "Pasta", "Dosa"],
    "price": [150, 280, 180]
})
```

```python
df = pd.concat([today_orders, yesterday_orders])

df = df.reset_index(drop=True)

print(df)
```
