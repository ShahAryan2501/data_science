# TASK - 4.

```python
import pandas as pd
import random

data = []

for i in range(1, 10001):
    data.append({
        "product_id": i,
        "product_name": "Product " + str(i),
        "category": random.choice(["Electronics", "Mobiles", "Fashion", "Home", "Beauty"]),
        "brand": random.choice(["Samsung", "Apple", "Sony", "Nike", "Puma"]),
        "price": random.randint(500, 100000),
        "rating": round(random.uniform(2.5, 5.0), 1),
        "reviews": random.randint(10, 50000)
    })

df = pd.DataFrame(data)

df.to_csv("flipkart_products.csv", index=False)

print("CSV file created successfully")
print("Total rows:", len(df))
```

```python
import pandas as pd

chunks = pd.read_csv("flipkart_products.csv", chunksize=2000)

for i, chunk in enumerate(chunks, start=1):
    print("Chunk", i, ":", len(chunk), "rows")
```
