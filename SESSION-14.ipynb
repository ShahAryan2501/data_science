```python
import csv
with open("matches.csv","w",newline = "") as f:
    writer = csv.writer(f)
    writer.writerow(["match_id","team-1","team-2","winner"])
    writer.writerow([1,"csk","rcb","csk"])
    writer.writerow([2,"mi","rr","rr"])
```

```python
with open("matches.csv","r") as f:
    reader = csv.reader(f)

    for i in reader:
        print(i[3])
```

```python
your_movies = [
    {
        "title": "3 Idiots",
        "year": 2009,
        "rating": 8.4
    },
    {
        "title": "Dangal",
        "year": 2016,
        "rating": 8.3
    },
    {
        "title": "Taare Zameen Par",
        "year": 2007,
        "rating": 8.3
    },
    {
    "title" : "saiyaara",
    "year" : 2025,
    "rating" : 9.2
    }
]
```

```python
import json
with open("createmovies.json","w") as f:
    json.dump(your_movies,f)
```

```python
with open("createmovies.json","r") as f:
    data = json.load(f)
    for movie in your_movies:
     print(movie["title"],"-",movie["rating"])
   
```
