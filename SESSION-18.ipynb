```python
import re

def check_date(text):
    if re.search(r"\d{2}/\d{2}/\d{4}", text):
        print("True")
    else:
        print("False")

paragraph = "Our school picnic was held on 15/08/2025. The next event is planned for 32/01/2026. Rahul's birthday is 07/11/2007."

check_date(paragraph)
```

```python
para = "I bought a T-shirt for Rs. 499, shoes for Rs. 2499, and a cap for Rs. 299. My friend purchased a bag for Rs. 1500."
digit = re.findall(r"\d+",para)
print(digit)
total = 0

for i in digit:
    total = total + int(i)
print("the sum of all this integer value is :",total)
```

```python
import re

text = "Contact us at aryan@gmail.com or support123@yahoo.com. You can also mail admin_01@outlook.com."

result = re.sub(r"[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}", "[hidden email]", text)

print(result)
```
